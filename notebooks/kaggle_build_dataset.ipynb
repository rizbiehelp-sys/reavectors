# ReaVector — build a Kaggle weights Dataset (run once)

Downloads the **complete diffusers** weights from Hugging Face into `/kaggle/working/reavector-models/`and writes a `reavector_models.json` mapping. Then you publish it as a **private Kaggle Dataset**, and theworker mounts it (no per-run download → no disk failure, no wasted quota).

**How to publish (after Run All):**
1. top-right **Save Version → Save & Run All**.
2. open the notebook **Output** tab → folder `reavector-models` → **... → New Dataset**.
3. name it `reavector-models`, visibility **Private**, **Create**.
4. send me the slug, e.g. `rizbiehelp/reavector-models`. I'll attach it to the worker.

Keep each run to **one big model** (Z-Image 12 GB or Wan 8 GB fit the ~20 GB output; FLUX 24 GB does not).

```python
import os, json
from huggingface_hub import snapshot_download

# pick ONE (or a small set) per run
MODELS = {
    "z-image-turbo": "Tongyi-MAI/Z-Image-Turbo",
    # "wan2.1-1.3b": "Wan-AI/Wan2.1-T2V-1.3B-Diffusers",
    # "sdxl-base-1.0": "stabilityai/stable-diffusion-xl-base-1.0",
}
ROOT = "/kaggle/working/reavector-models"
mapping = {}
for mid, repo in MODELS.items():
    dest = os.path.join(ROOT, mid)
    os.makedirs(dest, exist_ok=True)
    print("downloading", repo, "->", dest, flush=True)
    snapshot_download(repo_id=repo, local_dir=dest)
    mapping[mid] = mid
json.dump(mapping, open(os.path.join(ROOT, "reavector_models.json"), "w"), indent=2)
print("mapping:", mapping)
!du -sh /kaggle/working/reavector-models
!ls -1 /kaggle/working/reavector-models
```
